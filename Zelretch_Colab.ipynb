# Zelretch — Google Colab launcher

Run the cell below and paste a direct ZIP URL for your **Zelretch main project** when prompted. The Addons repository remains separate and is synchronized automatically at startup.

Developer: [Siam Chowdhury](https://github.com/ChowdhurySiam) · [@Ch0wdhury_Siam](https://t.me/Ch0wdhury_Siam)


```python
import os, pathlib, shutil, subprocess, urllib.request, zipfile

archive_url = os.environ.get("ZELRETCH_ARCHIVE_URL") or input("Zelretch ZIP URL: ").strip()
if not archive_url:
    raise ValueError("A direct Zelretch ZIP URL is required.")

workspace = pathlib.Path("/content/zelretch-runtime")
archive = pathlib.Path("/content/zelretch.zip")
shutil.rmtree(workspace, ignore_errors=True)
workspace.mkdir(parents=True, exist_ok=True)
urllib.request.urlretrieve(archive_url, archive)
with zipfile.ZipFile(archive) as bundle:
    bundle.extractall(workspace)

candidates = [p for p in workspace.iterdir() if p.is_dir() and (p / "main.py").exists()]
project = candidates[0] if candidates else workspace
subprocess.run(["python3", "-m", "venv", "venv"], cwd=project, check=True)
python = project / "venv" / "bin" / "python"
subprocess.run([str(python), "main.py", "--cli"], cwd=project, check=True)
```
